## Lambda Functions

- lambda function is nothing but a creating a function

```python
# create basic function

def add(x):
    return(x+10)

add(20)
```

```python
def add(x):
    summ=x+10
    return(summ)

add(20)
```

```python
def add(x):
    return(x+10)

add(20)


# how many arguments are present ?  : x
# what you are returning ?  : x+10
# # formate :    lambda <argument>:<output>
```

```python
# formate :   """"""" lambda <argument>:<output>""""""""

add=lambda x:x+10
add(20)
```

```python
 # implements using lambda 

def square(x):
    return(x*x)
  
```

```python
square=lambda a:a*a
square(10)
```

```python
 # implements using lambda 

def cube(x):
    return(x*x*x)
```

```python
cube=lambda b:b*b*b
cube(2)
```

$Two$ $arguments$

```python
def add(a,b):
    return(a+b)

add(20,30)


# how many arguments are present ?  : a,b
# what you are returning ?  : a+b

# # formate :    lambda <argument1>,<argument2>:<output>
```

```python
add=lambda a,b:a+b
add(20,30)
```

```python
# implement average of 3 numbers using lambda

average=lambda a,b,c:(a+b+c)/3
average(20,30,40)
```

```python
# implement average make c as default parameter

average=lambda a,b,c=30:round((a+b+c)/3,2)
average(20,30)
```

$if - else$

```python
# create a function for finding grater number between two numbers

def greater(n1,n2):
    if n1>n2:
        return(n1)

    else:
        return(n2)


greater(100,200)

# how many arguments ? : n1,n2 .... 2 arguments
# what you are returning ?   : n1>n2 or n2>n1
```

```python
# store above code output in ''"list"''

def greater(n1,n2):
    if n1>n2:
        return(n1)

    else:
        return(n2)

l1=[]
l1.append(greater(100,200))
l1
```

```python
# 2nd method
l1=[]
def greater(n1,n2):
    if n1>n2:
        l1.append(n1)

    else:
        l1.append(n2)


greater(100,200)
l1
```

```python
l1=[ n1 if n1>n2 else n2 for n1,n2 in [(100,200)]]
print(l1)
```

```python
# using lambda function

greater=lambda n1,n2:n1 if n1>n2 else n2
greater(30,70)
```

```python
list1=['hyd','mumbai','chennai']                                    # 34 th class
# output: ['Hyd','Mumbai','Mhennai'] 

# m-1: use append method 

list2=[]
for i in list1:
    list2.append(i.capitalize())

print(list2)


# M-2: use list comprhension

list3=[i.capitalize() for i in list1]
list3
```

```python
# now using lambda function
# lambda <arguments>: <outputs>,<iterator>

list1=['hyd','mumbai','chennai']
lambda i:i.capitalize(),list1
map(lambda i:i.capitalize(),list1)
list(map(lambda i:i.capitalize(),list1))

# map input and output using map key word
```

```python
list1=[1,2,3,4,5]
# apply the square == output: [1,4,9,16,25]

lambda i:i*i,list1
map(lambda i:i*i,list1)
list(map(lambda i:i*i,list1))
```

```python
list1=[1,2,3]
list2=[11,22,33]

# [12,24,36]

list(map(lambda i,j:i+j,list1,list2))
tuple(map(lambda i,j:i+j,list1,list2))
str(map(lambda i,j:i+j,list1,list2))
```

```python
list1=['h#d','mum#ai','chennai']
# ['h#d','mum#ai','chennai']

list2=[]
for i in list1:
    if '#' in i:
        list2.append(i)
print(list2)


# list comperhension

[i for i in list1 if '#' in i]

```

```python
#  lambda <argument>: <condition>,<iterator>

list1=['h#d','mum#ai','chennai']
list(map(lambda i:'#' in i,list1))

# in this case it is giving like this output
# for this we will use """"filter"""" insted of """"map""""
```

```python
# using filter

list1=['h#d','mum#ai','chennai']
list(filter(lambda i:'#' in i,list1))
```
