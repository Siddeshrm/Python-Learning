```python
# special function . this is a piece of code which belongs to ""count_div()"" 

def condition(num0,i,n_count,nd_count):
    if num0%i==0:
        print(f"{num0} is divisible by {i} ")
        n_count+=1
        
    
    else:
        print(f"{num0} is not divisible by {i} ")
        nd_count+=1

    return(n_count,nd_count)
```

```python
# in this code we are calling above code using function ==== condition    ################################################

def count_div2():
    n_count=0
    nd_count=0
    num0 = eval(input("which number u want to identify"))
    for i in range(1,num0+1):
        n_count,nd_count=condition(num0,i,n_count,nd_count)            # calling above special function
    return(n_count,nd_count)
    
val4,val5=count_div2()
print("the numbers of divisors are:",val4)
print("the numbers of divisors are:",val5)
```

**Patterns**

```python
# 1
for i in range(stop):

- in the bracket if we have only single value that is consider as stop value
- the default start value is = 0
- python index always start with :0
- if direction sign is not mentioned : default i will be increment (+) sign
- if direction is positive side then end = stop-1
```

```python
for i in range(10):
    print(i)

# start =0
# direction=+
# end or stop = 10-1 = 9
```

 **pattern 2**

for i in range(start, stop):

- start, start of the loop, initial point
- direction sign not mentioned: default +ve direction
- end = stop-1

```python
for i in range(10, 30):
    print(i,end=" ")
```

**pattern- 3**

for i in range(start, stop, step):

- start, start of the loop, initial point
- direction : what is sign of step value, that is direction

      - direction will not decided by start and stop value
      - direction will provided by step value
- if step size is positive direction

                - end = stop-1
- if step size is negative direction

                - end=stop+1

```python
for i in range(2,20,2):
    print(i,end=" ")
```

```python
for i in range(2,-20,-2):
    print(i,end=" ")
```
