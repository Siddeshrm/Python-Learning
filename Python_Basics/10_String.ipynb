```python
# WAP count the number of vowels in a string
# string= 'hai how are you and how do you do"

count=0
string="hai how are you"
for i in string:
    if i in 'aeiou':                                 # new method used ""in"" operator for ""if""
        count=count+1

print(f"number of vowels are {count}") 
```

```python
# WAP count the number of unique vowels in a string


string="hai how are you"
vowels=''
for i in string:
    if i in 'aeiou' and i not in vowels:
        vowels=vowels+i                        # new method used ""in"" operator for ""if""
        

print("Number of unique vowels:", len(vowels))
```
