**List comprehension**

```python
# writing a code in single line === it will speed up the process

list1=[1,2,3,4,5]  # input

output=[]
for i in list1:
    output.append(i*i)
output

# step-1: in above code intialized the list output=[]
# step-2: iterate the loop
# step-3: you save the output

output1=[i*i for i in list1]       # this is called List comprehension
output1
# [output> <for loop>]
```

```python
# list comprehension

output1=[i.capitalize() for i in list2]
output1
```

```python
# list comprehension with if condition


list3=['h#d','beng#luru','delhi']
# output=['h#d','beng#luru']
output=[]
for i in list3:
    if '#' in i:
        output.append(i)
output


#o/p= [ <which output we want ,that only we have to write>   <for loop>   <if condition> ]


output1=[i for i in list3 if '#' in i]
output1
```

```python
list4=['abcd','abc','ac','a',]
# elements less than 3
# ['ac','a']
for i in list4:
    if len(i)<3:
        print(i)


        
# now code using list comperhension

output1=[i for i in list4 if len(i)<3]
output1
```

```python
# if - else condition
list1 =[1,2,3,4,5,6]
output=[]
for i in list1:
    if i%2==0:
        output.append(f"{i} even")

    else:
        output.append(f"{i} odd")

output




# now code using list comperhension

#  [,if_output>   <if_condition>    else <else_output>     <for_loop>]


output1=[(f'{i} is even') if i%2==0  else (f'{i} is odd') for i in list1]
output1
```

```python
# find largest number in list without using max method

l = [100,-300,200,-500,-600,150]
maxi=0
for i in l:
    if maxi<i:
        maxi=i
        print(maxi)

print(f"maximum number is {maxi}")
```

```python
# another method

l = [100,300,200,500,600,150]
maxi=l[0]     # assigining 0 th index 
for element in l:
    if element> maxi:
        maxi=element
        print(maxi)

print(f"maximum number is {maxi}")
```
