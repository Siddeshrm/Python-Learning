```python
for cat in df.category:
    print(cat)      # it will print "GAME ACTION"
```

```python
for cat in df.category:
    print(cat.split(' ')[1]) # This will split 'GAME'  ' ACTION'
                                       #        0        1    INDEX
```

```python
text=" ".join(cat.split()[1] for cat in df.category)  # after splitting join will join that splitted names
text
```

```python
from wordcloud import WordCloud
wordcloud = WordCloud(height=2000, width=2000, background_color='black')
wordcloud = wordcloud.generate(" ".join(cat.split()[1] for cat in df.category))   # u can use like wordcloud.generate(text)
plt.imshow(wordcloud)
plt.title("Most commom words in positive customer comments")
plt.axis('off')
plt.show()
```
