$Method-2:np.where$

```python
file_path="C:\\DS_Py\\Visadataset.csv"
visa_df=pd.read_csv(file_path)
visa_df.head(5)
```

- np.where() is applicable for binary conditions
- which means it is applicable only for two labels
- np.where(condition,Truevalue,Falsee_value)
- for Example: case_status has 2 labels
- Certified==True, Replace certified values with 0
- False value: Replace all denied values with 1

```python
con=visa_df['case_status']=='Certified'
visa_df['case_status']=np.where(con,0,1)
```

$LabelEncoder$

- LableEncoder is a method from sklearn
- Under sklearn we have sub modules
- One of the submodule: preprocessing
- any sklearn packages we ahve only 3 steps

- step-1: read the package
- step-2: save the package
- step-3: Apply fit transform

```python
file_path="C:\\DS_Py\\Visadataset.csv"
visa_df=pd.read_csv(file_path)
print(visa_df['continent'].head(10))

# step-1
from sklearn.preprocessing import LabelEncoder

# step-2
le=LabelEncoder()
le

# step-3
visa_df['case_status']=le.fit_transform(visa_df['case_status'])
visa_df['continent']=le.fit_transform(visa_df['continent'])
print(visa_df['case_status'].head(10))
print(visa_df['continent'].head(10))
```

**One hot encoder**

- One hot encoder: at a time only one will be ON(1/True), others are OFF(0/False)
- Suppose case status has two unique lables
- Certified and denied
- in one hot encoder it will create 2 new columns

|case_status| Certified|denied|
|-----------|----------|------|
|certified|1|0|
|denied |0|1|

**Advantage**
  
- one hot encoder new columns are orthogonal each other
- Orthogonality means 90 degrees phase shift
- 90 degree phase shift means perpendicular each other
- perpendicular means independent each other
- Independent means no realtion
- It is very important property that is avoid realtion between variables

**Draw back**

- Assume that you have 100 unique lables for a column
- So it will create 100 new columns
- To process 100 columns we require more ""memory, more time, more hardware""
- This is called **Curse of Dimensionality**
