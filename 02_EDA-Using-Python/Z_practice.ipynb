**seabron, pi-chart           49 th class practice**

```python
# reading file normal step

file_path="C:\\DS_Py\\Visadataset.csv"
education=pd.read_csv(file_path)
education

# using seborn plot the graph
# sns.countplot(data=education,x='education_of_employee',color='orange')       # this will not give in order

# above graph is not in order
# lables=['High School','Masters','Bachelors','Doctorate']                     # this is manually given order
labels=education['education_of_employee'].value_counts().keys()
sns.countplot(data=education,x='education_of_employee',order=labels,color='orange')
plt.title("Bar plot")
plt.show()
```

```python
# pi-chart

keys=education['education_of_employee'].value_counts(normalize=True).keys()
values=education['education_of_employee'].value_counts(normalize=True).values

pd.DataFrame(zip(keys,values),columns=['Education','Relative freq'])


# now we will plot the graph
plt.pie(x=values,
        labels=keys,
        autopct="%0.1f%%",
        startangle=180,
        shadow=True,
        radius=2,
        explode=[0.4,0.4,0.4,0.4],
        counterclock=False
        )
plt.show()
```
