$Empirical-rule$

<img src="https://1.bp.blogspot.com/-RoWUojEiszE/X7ZYOmIB1LI/AAAAAAAAAJ4/x0eRiLJvKhM0HoHNpng1sL9SH7Bd4RDFgCLcBGAsYHQ/s2508/emperic%2Brule.png" id="img" style="width: 496px; height: 248px;">

**when data follows a normal distribution**
- u-1*sigma to u+1*sigma : 68%
- u-2*sigma to u+2*sigma : 95%
- u-3*sigma to u+3*sigma : 99.7%

```python
############# 68% #############
val_minus_1=round(wage_mean-1*wage_std,2)
val_plus_1=round(wage_mean+1*wage_std,2)

############# 95% #############
val_minus_2=round(wage_mean-2*wage_std,2)
val_plus_2=round(wage_mean+2*wage_std,2)

############ 99.7 #############
val_minus_3=round(wage_mean-3*wage_std,2)
val_plus_3=round(wage_mean+3*wage_std,2)

print(val_minus_1,val_plus_1,val_minus_2,val_plus_2,val_minus_3,val_plus_3)
```

- 68% of observation have values b/w [21639.87 127271.75]
- 95% of observation have values b/w [-31176.07 180087.69]
- 99.7% of observation have values b/w [-83992.01 232903.63]

```python
# for 68%
con1=visa_df["prevailing_wage"]>val_minus_1
con2=visa_df["prevailing_wage"]<val_plus_1
print(len(visa_df[con1&con2]))
len(visa_df[con1&con2])/len(visa_df)
```

```python
# for 95%
con3=visa_df["prevailing_wage"]>val_minus_2
con4=visa_df["prevailing_wage"]<val_plus_2
print(len(visa_df[con3&con4]))
len(visa_df[con3&con4])/len(visa_df)
```

```python
# for 99.7%
con5=visa_df["prevailing_wage"]>val_minus_3
con6=visa_df["prevailing_wage"]<val_plus_3
print(len(visa_df[con5&con6]))
len(visa_df[con5&con6])/len(visa_df)
```
