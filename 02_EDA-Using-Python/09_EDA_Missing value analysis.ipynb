**Missing value analysis**

```python
import pandas as pd                                                 # 55 th class
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# np.nan      Recommended
dict1={'Names':['Ram','Sita','Laxman',np.nan],
       'Age':[30,31,np.nan,33],
       'City':[np.nan,'Hyd','Pune','Chennai']}
df=pd.DataFrame(dict1)
df
```

```python
#None= is not recommended
dict1={'Names':['Ram','Sita','Laxman',None],
       'Age':[30,31,None,33],
       'City':[None,'Hyd','Pune','Chennai']}
pd.DataFrame(dict1)
```

```python
# Null
dict1={'Names':['Ram','Sita','Laxman','Null'],
       'Age':[30,31,'Null',33],
       'City':['Null','Hyd','Pune','Chennai']}
pd.DataFrame(dict1)
```

$Method-2$

**Fill the random value by using column wise**

```python
dict1={'Names':['Ram','Sita','Laxman',np.nan],
       'Age':[30,31,np.nan,33],
       'City':[np.nan,'Hyd','Pune','Chennai']}
df=pd.DataFrame(dict1)
df
```

```python
df['Names'].fillna("Raheem",inplace=True)
df['Age'].fillna(32,inplace=True)           # based on data types fill the values 
df['City'].fillna("Blr",inplace=True)
df
```

$Method-3$

- pad
- bfill
- backfill
- ffill

```python
dict1={'Names':['Ram','Sita','Laxman',np.nan],
       'Age':[30,31,np.nan,33],
       'City':[np.nan,'Hyd','Pune','Chennai']}
df=pd.DataFrame(dict1)
df
```

```python
print("======origina=======")
print(df)

print("======pad========")
print(df.fillna(method='pad'))

print("======ffill========")
print(df.fillna(method='ffill'))

print("======bfill========")
print(df.fillna(method='bfill'))

print("======backfill========")
print(df.fillna(method='backfill'))
```

```python
# axis=1
print("======origina=======")
print(df)

print("======pad========")
print(df.fillna(method='pad'))

print("======pad========")
print(df.fillna(method='pad',axis=1))
```

$Method-4$:
- Mean
- Median
- Mode

```python
########### read data again #########

dict1={'Names':['Ram','Sita','Laxman',np.nan],
       'Age':[30,31,np.nan,33],
       'City':[np.nan,'Hyd','Pune','Chennai']}
df=pd.DataFrame(dict1)
df
```

```python
mean_age=df['Age'].mean()
df['Age'].fillna(mean_age)
```

```python
median_age=df['Age'].median()
df['Age'].fillna(median_age)
```

- mode is used for categorical data
- Mean and median is used for numerical data
