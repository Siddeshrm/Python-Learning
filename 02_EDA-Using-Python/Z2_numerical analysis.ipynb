$Emperica-rule$

```python
# when data follows a normal distribution**
# u-1*sigma to u+1*sigma : 68%
# u-2*sigma to u+2*sigma : 95%
# u-3*sigma to u+3*sigma : 99.7%

mean=round(visa_df['prevailing_wage'].mean(),2)
std=round(visa_df['prevailing_wage'].std(),2)


############ 68% #########
minus1=round(mean-1*std,2)
plus1=round(mean+1*std,2)

########### 95% #########
minus2=round(mean-2*std,2)
plus2=round(mean+2*std,2)

########## 99.7% #########
minus3=round(mean-3*std,2)
plus3=round(mean+3*std,2)

print(minus1,plus1,minus2,plus2,minus3,plus3)
```

```python
# now we will check the condition

####### 68% #######
c1=visa_df['prevailing_wage']>minus1
c2=visa_df['prevailing_wage']<plus1
print(len(visa_df[c1&c2]))
print(len(visa_df[c1&c2])*100/len(visa_df))

####### 95 ##########
c3=visa_df['prevailing_wage']>minus2
c4=visa_df['prevailing_wage']<plus2
print(len(visa_df[c3&c4]))
print(len(visa_df[c3&c4])*100/len(visa_df))

####### 99.7 ##########
c5=visa_df['prevailing_wage']>minus3
c6=visa_df['prevailing_wage']<plus3
print(len(visa_df[c5&c6]))
print(len(visa_df[c5&c6])*100/len(visa_df))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

**outlair analysis**

```python
file_path="C:\\DS_Py\\Visadataset.csv"
visa_df=pd.read_csv(file_path)
visa_df.head(5)
```

```python
# box plot

plt.boxplot(visa_df['prevailing_wage'],vert=False)
plt.show()
```

```python
# finding outliers

q1=round(np.quantile(visa_df['prevailing_wage'],0.25),2)
q2=round(np.quantile(visa_df['prevailing_wage'],0.50),2)
q3=round(np.quantile(visa_df['prevailing_wage'],0.75),2)

# step-2: calculating IQR
IQR=q3-q1

# step-3: calculate UB and LB
UB=q3+1.5*IQR
LB=q1-1.5*IQR
print(UB)
print(LB)

# step-3: conditions > UB and <LB

con1=visa_df['prevailing_wage']>UB
con2=visa_df['prevailing_wage']<LB

# step-4: 
outliers=visa_df[con1|con2]
print("length of outliers is",len(outliers))
outliers
```

**remove the ouliers**
- 427 outliers re there, remove that from dataframe

```python
# in this small changes become ""non_outliers_dataset""

q1=round(np.quantile(visa_df['prevailing_wage'],0.25),2)
q2=round(np.quantile(visa_df['prevailing_wage'],0.50),2)
q3=round(np.quantile(visa_df['prevailing_wage'],0.75),2)

# step-2: calculating IQR
IQR=q3-q1

# step-3: calculate UB and LB
UB=q3+1.5*IQR
LB=q1-1.5*IQR

# step-3: conditions > UB and <LB

con1=visa_df['prevailing_wage']<UB                  # change from >  to <
con2=visa_df['prevailing_wage']>LB                  # change from < to >

non_outliers=visa_df[con1&con2]                     # change from | to &
print("total number of outliers are",len(non_outliers))
non_outliers
```

```python
# now plot the graphs

plt.subplot(2,2,1)
plt.title("non outlier data")
plt.hist(non_outliers['prevailing_wage'],bins=40)


plt.subplot(2,2,2)
plt.title("orignal data")
plt.hist(visa_df['prevailing_wage'],bins=40)

####### boxplot ##################

plt.subplot(2,2,3)
plt.title("non outlier data")
plt.boxplot(non_outliers['prevailing_wage'])

plt.subplot(2,2,4)
plt.title("non outlier data")
plt.boxplot(visa_df['prevailing_wage'])
plt.show()
```
