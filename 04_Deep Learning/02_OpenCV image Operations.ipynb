```python
# Traslation matrix the first row moves pixels horizontally and the second row moves in vertically
M = np.float32([[1, 0, 50], [0, 1, 120]])
M
```

```python
shifted = cv2.warpAffine(image, M, (image.shape[1], image.shape[0]))
plt.imshow(shifted)
```

```python
# change the pixel directions to move image left and up

# [1, 0, -50]  [ row, column, number of pixels ]

M = np.float32([[1, 0, -60], [0, 1, -60]])
shifted = cv2.warpAffine(image, M, (image.shape[1], image.shape[0]))
plt.imshow(shifted)
```

```python
(height, width, channels)=image.shape
centre=(width//2, height//2)
centre
```

```python
M = cv2.getRotationMatrix2D(centre,45,0.8)       # here 45 degree angle and 0.8 is scaling means zooming
rotated = cv2.warpAffine(image, M, (width, height))
plt.imshow(rotated)
```

**Resizing**

```python
# Here we resizing settings width to 150 and scale the height accordingly
r = 150.0/ image.shape[1]
dim=(150,int(image.shape[0] * r))
resized = cv2.resize(image, dim, interpolation = cv2.INTER_AREA)
plt.imshow(resized)
```

**Flipping**

```python
# Flipping across y - axis
flipped = cv2.flip(image, 1)
plt.imshow(flipped)
```

```python
flipped = cv2.flip(image, 0)
plt.imshow(flipped)
```

```python
flipped = cv2.flip(image, -1)
plt.imshow(flipped)
```
