**Steps to capture a video**:

- Use cv2. videoCapture() to get a video capture object for the camera.
- Set up an infinite while loop and use the read() method to read the frames using the above created object.
- Use cv2.iamshow() method to show the frames in the video.
- Breaks the loop when the user clicks a specific key.

```python
import cv2
cv2.imread
cv2.imshow
cv2.imwrite
cv2.VideoCapture(0)      # It will open the camera laptop. capture our videos
```

```python
# import the opencv library
import cv2

# define a video capture object
vid = cv2.VideoCapture(0)

while True:
    # Capture the video frame
    # by frame
    ret, frame = vid.read()
    # ret = boolean value
    # ret = True, the frame is captured
    # ret = False, the frame is not capture
    print(ret)
    # Display the resulting frame
    cv2.imshow('frame', frame)
    # the 'q' button is set as the
    # quitting button you may use any
    # desired button of your choice
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# After the loop release the cap object
vid.release()
# Destroy all the windows
cv2.destroyAllWindows()
```

```python
# Importing all necessary libraries
import cv2
import os

# Read the video from specified path
cam =cv2.VideoCapture("C:\\DS_Py\\04_Deep Learning\\SIddesh.mp4")

try:
    # creating a folder named data
    if not os.path.exists('data'):
        os.makedirs('data')

# if not created then raise error
except OSError:
    print('Error: Creating directory of data')

# frame
currentframe = 0

while True:
    # reading from frame
    ret, frame = cam.read()
    if ret:
        # if video is still left continue creating images
        name = './data/frame'+str(currentframe)+'.jpg'
        print('Creating...' + name)
        # writing the extracted images
        cv2.imwrite(name, frame)
        # increasing counter so that it will
        # show how many frames are created
        currentframe+=1
    else:
        break

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

        
# Release all space and windows once done
cam.release()
cv2.destroyAllWindows()
```

**Write a text on video**

```python
import cv2
vid = cv2.VideoCapture(0)

while True:
    ret, frame = vid.read()
    print(ret)
    cv2.imshow('frame', frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

vid.release()
cv2.destroyAllWindows()
```

```python
import cv2

cap = cv2.VideoCapture(0)

while(True):
    # capture frames in the video
    ret, frame = cap.read()

    # describe the type of font
    # to be used
    font = cv2.FONT_HERSHEY_SIMPLEX

    # Use putText() method for
    # inserting text on video R=0, G=255,B=255
    cv2.putText(frame,
                'hello siddesh',
                (50,50),
                font,
                1,
                (0, 255, 255),
                2,
                cv2.LINE_4)
    # Display the resulting frame
    cv2.imshow('video',frame)

    # creating 'q' as the quit
    # button for the video
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# release the cap object
cap.release()
# close all windows
cv2.destroyAllWindows()
```
