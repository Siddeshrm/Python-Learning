**We use frontalface model to identify face**

```python
import cv2
img=cv2.imread(r"C:\\DS_Py\\03_ML Python\\Benjamin-Franklin.jpg",cv2.COLOR_RGB2GRAY)
face_cascade=cv2.CascadeClassifier(r"C:\DS_Py\04_Deep Learning\haarcascade_frontalface_default.xml")
faces=face_cascade.detectMultiScale(img,1.1,4)
print(faces)
for (x,y,w,h) in faces:
    print(x,y)
    print(w,h)
    cv2.rectangle(img,(x,y),(x+w,y+h),(0,0,255),3)    # BGR red is ON
```

```python
cv2.imshow('image',img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

**Video**:

```python
cap=cv2.VideoCapture(0)
while cap.isOpened():      # True or False
    has_frame,frame=cap.read()      # cap.read will give twoq
    imag=cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
    faces=face_cascade.detectMultiScale(imag,1.1,4)
    for (x,y,w,h) in faces:
        cv2.rectangle(imag,(x,y),(x+y,y+w),(255,0,0),3)

    cv2.imshow('image',imag)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
        
cap.release()
cv2.destroyAllWindows()
```
